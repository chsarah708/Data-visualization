```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load CSV from the same folder as this Python file
from pathlib import Path

folder = Path(__file__).resolve().parent
df = pd.read_csv(folder / "datasaurus.csv")

# Calculate statistics for each dataset
grouped = df.groupby("dataset")

summary = grouped.agg(
    mean_x=("x", "mean"),
    mean_y=("y", "mean"),
    std_x=("x", "std"),  # Sample standard deviation (n - 1)
    std_y=("y", "std"),
)

summary["correlation"] = pd.Series({
    name: group["x"].corr(group["y"])
    for name, group in grouped
})

print("\nSUMMARY STATISTICS\n")
print(summary.round(4).to_string())
summary.to_csv(folder / "summary_statistics.csv")

# 2. Conclusion based only on the table
print("""
CONCLUSION FROM THE TABLE:
All 13 datasets have almost identical means and standard deviations.
Their correlations are close to zero, indicating very little linear
association between x and y. Based only on these statistics, we might
expect the datasets to look similar.
""")

# 3. Plot all 13 datasets using identical scales
fig, axes = plt.subplots(
    4, 4,
    figsize=(14, 14),
    sharex=True,
    sharey=True,
)

for ax, (name, group) in zip(axes.flat, grouped):
    ax.scatter(group["x"], group["y"], s=15, color="teal")
    ax.set_title(name)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_xlim(-2, 102)
    ax.set_ylim(-2, 102)
    ax.set_aspect("equal", adjustable="box")
    ax.grid(alpha=0.2)

# Hide unused panels
for ax in axes.flat[grouped.ngroups:]:
    ax.set_visible(False)

fig.suptitle("Datasaurus: Similar Statistics, Different Shapes", fontsize=18)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig(folder / "datasaurus_scatter_plots.png", dpi=300)

# 4. Findings and answer
print("""
FINDINGS AFTER VISUALIZATION:
The datasets have very different shapes, including a dinosaur, circle,
star, clusters, and horizontal, vertical, and diagonal lines.
These patterns are hidden by their nearly identical summary statistics.
A correlation near zero does not mean there is no relationship;
there may be a strong nonlinear pattern.

ANSWER:
No, having 13 datasets with 142 points each does not make these summary
statistics sufficient to describe the data. Larger representative samples
can improve estimation precision, but means, standard deviations, and
correlations still cannot describe every shape, cluster, or nonlinear
relationship. We should use summary statistics together with visualization.
""")

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Load the penguins dataset directly from the internet
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)

# Standardize sex labels while preserving missing values
df["sex"] = df["sex"].astype("string").str.lower()

species_order = ["Adelie", "Chinstrap", "Gentoo"]
colors = {
    "Adelie": "#0072B2",
    "Chinstrap": "#D55E00",
    "Gentoo": "#009E73",
}

# ---------------------------------------------------------
# FIND MISSING VALUES FIRST
# ---------------------------------------------------------
measurements = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g",
]

print("Dataset shape:", df.shape)
print("\nMissing values per column:")
display(df.isna().sum().to_frame("missing_count"))

print("\nPenguins missing all four measurements:")
display(df.loc[df[measurements].isna().all(axis=1)])

print("\nPenguins missing sex:")
display(df.loc[df["sex"].isna()])

print("""
MISSING-DATA DECISION:
Do not impute measurements or guess sex.
For each chart, exclude only rows missing variables needed by that chart.
Keep penguins with missing sex when sex is not used.
The two penguins missing all measurements are included among the
11 penguins missing sex, so these are not 13 separate penguins.
""")

# Prepare data separately for each task
mass_data = df.dropna(subset=["species", "body_mass_g"]).copy()
bill_data = df.dropna(
    subset=["species", "bill_length_mm", "bill_depth_mm"]
).copy()
sex_data = df.dropna(
    subset=["species", "sex", "body_mass_g"]
).copy()

def missing_note(data, uses_sex=False):
    excluded = len(df) - len(data)
    if uses_sex:
        return (
            f"n = {len(data)}; excluded {excluded} rows missing mass or sex "
            "(including the 2 missing all measurements). No imputation."
        )
    retained = data["sex"].isna().sum()
    return (
        f"n = {len(data)}; excluded {excluded} rows missing required measurements; "
        f"retained {retained} with missing sex. No imputation."
    )

# Use identical 250 g bins for both distribution charts
bin_width = 250
bins = np.arange(
    np.floor(mass_data["body_mass_g"].min() / bin_width) * bin_width,
    np.ceil(mass_data["body_mass_g"].max() / bin_width) * bin_width
    + bin_width,
    bin_width,
)

# ---------------------------------------------------------
# 1. OVERALL BODY-MASS DISTRIBUTION
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(
    mass_data["body_mass_g"],
    bins=bins,
    color="#577590",
    edgecolor="white",
)

ax.set(
    title="Body-mass distribution: all penguins",
    xlabel="Body mass (g)",
    ylabel="Number of penguins",
)
ax.grid(axis="y", alpha=0.2)
ax.set_axisbelow(True)

fig.text(
    0.5, 0.02,
    missing_note(mass_data),
    ha="center", fontsize=9,
)
fig.tight_layout(rect=[0, 0.07, 1, 1])
plt.show()

print("""
1. DISTRIBUTION:
The distribution is roughly bimodal, with lighter and heavier groups,
rather than one simple symmetric hump. Smaller peaks depend on bin width.
""")

# ---------------------------------------------------------
# 2. BODY-MASS DISTRIBUTION SPLIT BY SPECIES
# ---------------------------------------------------------
fig, axes = plt.subplots(
    1, 3,
    figsize=(15, 5),
    sharex=True,
    sharey=True,
)

for ax, species in zip(axes, species_order):
    group = mass_data[mass_data["species"] == species]

    ax.hist(
        group["body_mass_g"],
        bins=bins,
        color=colors[species],
        edgecolor="white",
    )
    ax.set_title(f"{species} (n = {len(group)})")
    ax.set_xlabel("Body mass (g)")
    ax.grid(axis="y", alpha=0.2)
    ax.set_axisbelow(True)

axes[0].set_ylabel("Number of penguins")
fig.suptitle("Body-mass distribution by species", fontsize=16)

fig.text(
    0.5, 0.02,
    missing_note(mass_data),
    ha="center", fontsize=9,
)
fig.tight_layout(rect=[0, 0.08, 1, 0.93])
plt.show()

print("""
2. SPECIES EXPLAIN THE SHAPE:
Adelie and Chinstrap penguins have overlapping, lighter body masses.
Gentoo penguins are generally heavier. Mixing these species produces
the broad lighter and heavier groups seen in the overall histogram.
Species explains much of the shape, although sex also contributes
to variation within species.
""")

# ---------------------------------------------------------
# 3. BILL LENGTH VS BILL DEPTH: SIMPSON'S PARADOX
# ---------------------------------------------------------
overall_r = bill_data["bill_length_mm"].corr(
    bill_data["bill_depth_mm"]
)

correlations = {"All penguins": overall_r}

fig, ax = plt.subplots(figsize=(11, 7))

for species in species_order:
    group = bill_data[bill_data["species"] == species]
    x = group["bill_length_mm"].to_numpy()
    y = group["bill_depth_mm"].to_numpy()

    r = group["bill_length_mm"].corr(group["bill_depth_mm"])
    correlations[species] = r

    ax.scatter(
        x, y,
        color=colors[species],
        alpha=0.65,
        s=35,
        label=f"{species}: r = {r:.2f}",
    )

    # Within-species fitted line
    slope, intercept = np.polyfit(x, y, 1)
    x_line = np.linspace(x.min(), x.max(), 100)

    ax.plot(
        x_line,
        slope * x_line + intercept,
        color=colors[species],
        linewidth=2.5,
    )

# Overall fitted line
x_all = bill_data["bill_length_mm"].to_numpy()
y_all = bill_data["bill_depth_mm"].to_numpy()
slope, intercept = np.polyfit(x_all, y_all, 1)
x_line = np.linspace(x_all.min(), x_all.max(), 100)

ax.plot(
    x_line,
    slope * x_line + intercept,
    color="black",
    linestyle="--",
    linewidth=2.5,
    label=f"All penguins: r = {overall_r:.2f}",
)

ax.set(
    title="Simpson's paradox: pooled vs within-species relationships",
    xlabel="Bill length (mm)",
    ylabel="Bill depth (mm)",
)
ax.legend(title="Species and Pearson correlation")
ax.grid(alpha=0.2)

fig.text(
    0.5, 0.02,
    missing_note(bill_data),
    ha="center", fontsize=9,
)
fig.tight_layout(rect=[0, 0.07, 1, 1])
plt.show()

print("\nPearson correlations:")
display(
    pd.Series(correlations, name="correlation")
    .to_frame()
    .round(3)
)

print("""
3. SIMPSON'S PARADOX:
Across all penguins, bill length and bill depth are negatively related.
Within each species, they are positively related.

Species is associated with both measurements. Gentoo penguins tend
to have longer but shallower bills than Adelie penguins, while Chinstrap
penguins have relatively long and deep bills. Differences between
species pull the pooled trend downward, even though the trend within
each species slopes upward.

This reversal is called Simpson's paradox. Pooling different groups
can reverse the relationship found inside those groups. Neither
correlation by itself establishes causation.

Reference:
https://online.stat.psu.edu/stat100/Lesson06
""")

# ---------------------------------------------------------
# 4. AVERAGE BODY MASS BY SPECIES AND SEX
# ---------------------------------------------------------
summary = (
    sex_data.groupby(["species", "sex"])["body_mass_g"]
    .agg(mean_mass="mean", count="size")
)

means = (
    summary["mean_mass"]
    .unstack("sex")
    .reindex(index=species_order, columns=["female", "male"])
)
counts = (
    summary["count"]
    .unstack("sex")
    .reindex(index=species_order, columns=["female", "male"])
)

positions = np.arange(len(species_order))
width = 0.36

fig, ax = plt.subplots(figsize=(11, 7))

for sex, offset, color in [
    ("female", -width / 2, "#CC79A7"),
    ("male", width / 2, "#56B4E9"),
]:
    bars = ax.bar(
        positions + offset,
        means[sex],
        width=width,
        color=color,
        label=sex.capitalize(),
    )

    for bar, species in zip(bars, species_order):
        mean = means.loc[species, sex]
        count = int(counts.loc[species, sex])

        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 70,
            f"{mean:,.0f} g\nn = {count}",
            ha="center",
            va="bottom",
            fontsize=9,
        )

ax.set_xticks(positions)
ax.set_xticklabels(species_order)
ax.set(
    title="Average body mass by species and sex",
    xlabel="Species",
    ylabel="Mean body mass (g)",
)
ax.set_ylim(0, np.nanmax(means.to_numpy()) * 1.22)
ax.legend(title="Sex")
ax.grid(axis="y", alpha=0.2)
ax.set_axisbelow(True)

fig.text(
    0.5, 0.02,
    missing_note(sex_data, uses_sex=True),
    ha="center", fontsize=9,
)
fig.tight_layout(rect=[0, 0.08, 1, 1])
plt.show()

print("\nAverage body mass and sample sizes:")
display(summary.round(1))

print("""
4. COMPARISON:
Male penguins have higher average body mass than females within
each species. Gentoo penguins have the highest average body mass
for both sexes. This chart includes only penguins with recorded
body mass and sex; missing sex was not guessed.
""")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Load data directly from the internet
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
df = pd.read_csv(url)

df["year"] = df["model_year"] + 1900
df["origin"] = df["origin"].replace({
    "usa": "American",
    "europe": "European",
    "japan": "Japanese",
})

print("Total cars:", len(df))
print("\nMissing values:")
display(df.isna().sum().to_frame("missing_count"))

print("\nCars missing horsepower:")
display(df.loc[df["horsepower"].isna()])

print("""
MISSING-DATA DECISION:
Keep all 398 cars. Horsepower is not used in these charts, so its
six missing values do not require dropping rows or imputing values.
""")

note = (
    "All 398 cars retained; 6 missing horsepower values "
    "are irrelevant to these charts."
)

def finish(fig):
    fig.text(0.5, 0.015, note, ha="center", fontsize=9)
    fig.tight_layout(rect=[0, 0.06, 1, 1])
    plt.show()

yearly = (
    df.groupby("year", as_index=False)
    .agg(
        mean_mpg=("mpg", "mean"),
        mean_weight=("weight", "mean"),
        cars=("mpg", "size"),
    )
)

# ---------------------------------------------------------
# 1. DID CARS GET MORE FUEL-EFFICIENT?
# Chart: line chart of yearly mean MPG
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    yearly["year"],
    yearly["mean_mpg"],
    marker="o",
    color="#0072B2",
    linewidth=2.5,
)

ax.set(
    title="Average fuel efficiency by model year",
    xlabel="Model year",
    ylabel="Mean fuel efficiency (MPG)",
)
ax.set_xticks(yearly["year"])
ax.tick_params(axis="x", rotation=45)
ax.grid(alpha=0.2)

finish(fig)

start_mpg = yearly["mean_mpg"].iloc[0]
end_mpg = yearly["mean_mpg"].iloc[-1]

print(
    "1. JUSTIFICATION: A line chart shows how average fuel efficiency "
    "changes across ordered model years."
)
print(
    f"FINDING: Mean MPG changed from {start_mpg:.2f} in 1970 "
    f"to {end_mpg:.2f} in 1982."
)
print("These are different model-year groups, not the same cars tracked over time.")

# ---------------------------------------------------------
# 2. DOES FUEL EFFICIENCY DIFFER BY ORIGIN?
# Chart: box plots
# ---------------------------------------------------------
origins = ["American", "European", "Japanese"]
origin_colors = ["#0072B2", "#D55E00", "#009E73"]

origin_values = [
    df.loc[df["origin"] == origin, "mpg"].to_numpy()
    for origin in origins
]

fig, ax = plt.subplots(figsize=(10, 6))

boxes = ax.boxplot(
    origin_values,
    patch_artist=True,
    medianprops={"color": "black", "linewidth": 2},
)

for box, color in zip(boxes["boxes"], origin_colors):
    box.set_facecolor(color)
    box.set_alpha(0.65)

ax.set_xticks([1, 2, 3])
ax.set_xticklabels([
    f"{origin}\n(n = {len(values)})"
    for origin, values in zip(origins, origin_values)
])
ax.set(
    title="Fuel-efficiency distributions by origin",
    xlabel="Origin",
    ylabel="Fuel efficiency (MPG)",
)
ax.grid(axis="y", alpha=0.2)

finish(fig)

print(
    "\n2. JUSTIFICATION: Box plots compare typical fuel efficiency, "
    "spread, and possible outliers across the three origins."
)
display(
    df.groupby("origin")["mpg"]
    .agg(count="size", mean="mean", median="median")
    .round(2)
)
print(
    "This is an unadjusted comparison: differences in model year, "
    "weight, and other characteristics may contribute."
)

# ---------------------------------------------------------
# 3. RELATIONSHIP BETWEEN WEIGHT AND MPG
# Chart: scatter plot with a fitted linear trend
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(
    df["weight"],
    df["mpg"],
    color="#7B3294",
    alpha=0.55,
    s=35,
    edgecolors="none",
)

slope, intercept = np.polyfit(df["weight"], df["mpg"], 1)
weight_line = np.linspace(df["weight"].min(), df["weight"].max(), 100)

ax.plot(
    weight_line,
    slope * weight_line + intercept,
    color="black",
    linestyle="--",
    linewidth=2,
    label="Fitted linear trend",
)

ax.set(
    title="Heavier cars generally have lower fuel efficiency",
    xlabel="Weight (lb)",
    ylabel="Fuel efficiency (MPG)",
)
ax.legend()
ax.grid(alpha=0.2)

finish(fig)

print(
    "\n3. JUSTIFICATION: A scatter plot shows the direction, "
    "shape, and variation of the relationship between two numeric variables."
)
print(f"Pearson correlation: {df['weight'].corr(df['mpg']):.3f}")
print("Association does not establish causation.")

# ---------------------------------------------------------
# 4. DISTRIBUTION OF CYLINDER COUNTS
# Chart: bar chart
# ---------------------------------------------------------
cylinder_counts = df["cylinders"].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    cylinder_counts.index.astype(str),
    cylinder_counts.values,
    color="#009E73",
    edgecolor="white",
)

for bar, count in zip(bars, cylinder_counts.values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        count + 3,
        f"{count}\n({count / len(df):.1%})",
        ha="center",
        fontsize=10,
    )

ax.set(
    title="Number of cylinders across the fleet",
    xlabel="Number of cylinders",
    ylabel="Number of cars",
)
ax.set_ylim(0, cylinder_counts.max() * 1.2)
ax.grid(axis="y", alpha=0.2)
ax.set_axisbelow(True)

finish(fig)

print(
    "\n4. JUSTIFICATION: A bar chart directly compares frequencies "
    "of the discrete cylinder categories."
)

# ---------------------------------------------------------
# 5. DID CARS GET LIGHTER AS MPG IMPROVED?
# Initial chart: two time series using separate y-axes
# ---------------------------------------------------------
fig, ax_mpg = plt.subplots(figsize=(10, 6))
ax_weight = ax_mpg.twinx()

mpg_line, = ax_mpg.plot(
    yearly["year"],
    yearly["mean_mpg"],
    color="#0072B2",
    marker="o",
    linewidth=2.5,
    label="Mean MPG",
)

weight_line, = ax_weight.plot(
    yearly["year"],
    yearly["mean_weight"],
    color="#D55E00",
    marker="s",
    linewidth=2.5,
    label="Mean weight",
)

ax_mpg.set(
    title="Fuel efficiency and weight over model years",
    xlabel="Model year",
    ylabel="Mean fuel efficiency (MPG)",
)
ax_weight.set_ylabel("Mean weight (lb)")
ax_mpg.tick_params(axis="y", labelcolor="#0072B2")
ax_weight.tick_params(axis="y", labelcolor="#D55E00")
ax_mpg.set_xticks(yearly["year"])
ax_mpg.tick_params(axis="x", rotation=45)
ax_mpg.grid(alpha=0.2)
ax_mpg.legend(handles=[mpg_line, weight_line], loc="upper left")

finish(fig)

print(
    "\n5. JUSTIFICATION: Two time series on a shared time axis "
    "allow comparison of yearly average weight and fuel-efficiency trends."
)

start_weight = yearly["mean_weight"].iloc[0]
end_weight = yearly["mean_weight"].iloc[-1]

print(
    f"Mean weight changed from {start_weight:.1f} lb in 1970 "
    f"to {end_weight:.1f} lb in 1982."
)
print("""
INTERPRETATION:
Cars became lighter on average between the first and last model years,
while average MPG increased. Together with the weight–MPG scatter plot,
this is consistent with lower weight contributing to improved efficiency.
However, these charts cannot show how much of the improvement weight
explains or establish causation; other changes may also contribute.
""")

# ---------------------------------------------------------
# REDRAW THE WEAKEST CHART: QUESTION 5
# Replace dual axes with percentage changes from 1970
# ---------------------------------------------------------
yearly["mpg_change_pct"] = (
    yearly["mean_mpg"] / start_mpg - 1
) * 100

yearly["weight_change_pct"] = (
    yearly["mean_weight"] / start_weight - 1
) * 100

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    yearly["year"],
    yearly["mpg_change_pct"],
    color="#0072B2",
    marker="o",
    linewidth=2.5,
    label="Mean MPG",
)
ax.plot(
    yearly["year"],
    yearly["weight_change_pct"],
    color="#D55E00",
    marker="s",
    linewidth=2.5,
    label="Mean weight",
)

ax.axhline(0, color="grey", linestyle="--", linewidth=1)

ax.set(
    title="Redraw: MPG and weight changes relative to 1970",
    xlabel="Model year",
    ylabel="Percentage change from the 1970 mean (%)",
)
ax.set_xticks(yearly["year"])
ax.tick_params(axis="x", rotation=45)
ax.legend()
ax.grid(alpha=0.2)

finish(fig)

print("""
WHAT I CHANGED:
Question 5 was the weakest because independently scaled y-axes can
make the visual relationship depend on arbitrary axis choices.
I replaced them with one common percentage-change scale, using
1970 as the baseline. This makes proportional changes easier to
compare, although it no longer displays absolute MPG or weight.
""")

print(
    f"1970–1982: mean MPG changed by "
    f"{yearly['mpg_change_pct'].iloc[-1]:+.1f}%, "
    f"and mean weight changed by "
    f"{yearly['weight_change_pct'].iloc[-1]:+.1f}%."
)
```
